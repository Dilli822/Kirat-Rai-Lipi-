```python

# =====================================================
# Replace with our results from multiple seeds
# =====================================================

results = {
    "ViT-Small DINOv3": [25.90, 26.20, 25.30, 26.00, 25.80],
    "MobileNetV4 Small": [68.07, 67.80, 68.50, 67.90, 68.20],
    "MobileOne-S1": [62.65, 63.10, 62.00, 62.70, 62.80],
    "TinyNet-E": [60.84, 61.50, 60.20, 60.90, 61.00],
    "EfficientViT-M0": [57.83, 58.10, 57.20, 57.90, 58.00],
    "TinyViT-5M": [96.39, 96.60, 96.20, 96.50, 96.30],
    "EfficientFormer-L1": [97.59, 97.80, 97.50, 97.70, 97.60],
    "EdgeNeXt-XX-Small": [99.40, 99.50, 99.30, 99.40, 99.50],
    "EfficientFormerV2-S0": [99.40, 99.50, 99.40, 99.30, 99.50],
    "FastViT-T8": [99.40, 99.30, 99.50, 99.40, 99.50]
}

# =====================================================
# Compute Mean and Std
# =====================================================

summary = []

for model, scores in results.items():
    summary.append({
        "Model": model,
        "Mean": np.mean(scores),
        "Std": np.std(scores, ddof=1)
    })

df = pd.DataFrame(summary)

# Sort by mean accuracy
df = df.sort_values("Mean", ascending=True)

# =====================================================
# Plot
# =====================================================

plt.figure(figsize=(10,6))

plt.barh(
    df["Model"],
    df["Mean"],
    xerr=df["Std"],
    capsize=5
)

plt.xlabel("Top-1 Accuracy (%)")
plt.ylabel("Model")
plt.title("Model Comparison (Mean ± Std Across Multiple Seeds)")

plt.tight_layout()
plt.savefig("accuracy_comparison_mean_std.pdf", bbox_inches="tight")
plt.savefig("accuracy_comparison_mean_std.png", dpi=300, bbox_inches="tight")

plt.show()

# =====================================================
# Table for paper
# =====================================================

df["Result"] = df.apply(
    lambda x: f"{x['Mean']:.2f} ± {x['Std']:.2f}",
    axis=1
)

print(df[["Model", "Result"]])
```

```python

# ==========================================================
# Prepare Data
# ==========================================================

data_list = []

for name, metrics in model_data.items():
    data_list.append({
        "Model": name.replace("_", " ").title(),
        "Params_M": metrics["params"] / 1e6,
        "Accuracy": metrics["acc"],
        "Std": metrics["std"]
    })

df = pd.DataFrame(data_list)

# ==========================================================
# Figure Style
# ==========================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 14,
    "axes.labelsize": 20,
    "axes.titlesize": 22,
    "xtick.labelsize": 16,
    "ytick.labelsize": 16,
    "legend.fontsize": 14
})

fig, ax = plt.subplots(figsize=(16, 8))

# ==========================================================
# Color Mapping
# ==========================================================

norm = mcolors.Normalize(
    vmin=df["Accuracy"].min(),
    vmax=df["Accuracy"].max()
)

cmap = plt.get_cmap("RdYlGn")

# ==========================================================
# Scatter Plot
# ==========================================================

scatter = ax.scatter(
    df["Params_M"],
    df["Accuracy"],
    c=df["Accuracy"],
    cmap=cmap,
    s=220,
    edgecolor="black",
    linewidth=1,
    zorder=3
)

# ==========================================================
# Error Bars
# ==========================================================

ax.errorbar(
    df["Params_M"],
    df["Accuracy"],
    yerr=df["Std"],
    fmt="none",
    ecolor="dimgray",
    elinewidth=1.5,
    capsize=4,
    zorder=2
)

# ==========================================================
# Smart Label Placement
# ==========================================================

texts = []

for _, row in df.iterrows():
    texts.append(
        ax.text(
            row["Params_M"],
            row["Accuracy"],
            row["Model"],
            fontsize=12
        )
    )

adjust_text(
    texts,
    ax=ax,
    expand_points=(2.0, 2.0),
    expand_text=(1.5, 1.5),
    force_points=0.6,
    force_text=0.6,
    arrowprops=dict(
        arrowstyle="-",
        color="gray",
        lw=0.8,
        alpha=0.7
    )
)

# ==========================================================
# Formatting
# ==========================================================

ax.set_xscale("log")

ax.set_xlabel("Trainable Parameters (Millions, Log Scale)")
ax.set_ylabel("Top-1 Accuracy (%)")

ax.set_title(
    "Accuracy–Efficiency Trade-off of Lightweight Pretrained Models"
)

ax.set_ylim(20, 102)

ax.grid(
    True,
    linestyle="--",
    linewidth=0.7,
    alpha=0.4
)

# ==========================================================
# Colorbar
# ==========================================================

cbar = plt.colorbar(scatter, ax=ax, pad=0.02)
cbar.set_label("Top-1 Accuracy (%)")

# ==========================================================
# Save
# ==========================================================

plt.tight_layout()

plt.savefig(
    "accuracy_vs_parameters_benchmark.pdf",
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    "accuracy_vs_parameters_benchmark.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
